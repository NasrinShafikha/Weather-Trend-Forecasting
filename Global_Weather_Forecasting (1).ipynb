```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('GlobalWeatherRepository.csv')

# Display dataset information
print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
df.info()
```

```python

# 1. Check data types
print("DATA TYPES")
display(df.dtypes)

# 2. Check missing values
print("\nMISSING VALUES")
missing = df.isnull().sum()
print(missing[missing > 0])
print("Total missing values:", df.isnull().sum().sum())

# 3. Check duplicate records
print("\nDUPLICATE RECORDS")
print("Total duplicates:", df.duplicated().sum())

# 4. Check dataset statistics
print("\nSTATISTICAL SUMMARY")
display(df.describe().T)

# 5. Convert last_updated to datetime
df['last_updated'] = pd.to_datetime(df['last_updated'])

print("\nDate range:")
print("Start:", df['last_updated'].min())
print("End:", df['last_updated'].max())
```

```python

# ==========================================
# DATA CLEANING AND PREPROCESSING
# ==========================================

# Create a separate cleaned dataset
df_clean = df.copy()

# 1. Convert date column
df_clean['last_updated'] = pd.to_datetime(
    df_clean['last_updated'],
    errors='coerce'
)

# 2. Handle invalid sentinel values
# -9999 represents invalid/missing air-quality readings
sentinel_columns = [
    'air_quality_Carbon_Monoxide',
    'air_quality_Sulphur_dioxide'
]

for col in sentinel_columns:
    df_clean[col] = df_clean[col].replace(-9999, np.nan)

# 3. Handle missing values created by sentinel replacement
for col in sentinel_columns:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )

# 4. Verify cleaning
print("CLEANED DATASET SHAPE:", df_clean.shape)

print("\nRemaining Missing Values:",
      df_clean.isnull().sum().sum())

print("\nRemaining Duplicates:",
      df_clean.duplicated().sum())

print("\nDate Column Type:",
      df_clean['last_updated'].dtype)

print("\nCleaned Air Quality Values:")
display(df_clean[sentinel_columns].describe())
```

```python

# ==========================================
# OUTLIER DETECTION USING IQR
# ==========================================

numerical_cols = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

outlier_results = []

for col in numerical_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_limit) |
        (df_clean[col] > upper_limit)
    ]

    outlier_results.append({
        'Feature': col,
        'Lower Limit': round(lower_limit, 2),
        'Upper Limit': round(upper_limit, 2),
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

outlier_df = outlier_df.sort_values(
    by='Outlier Count',
    ascending=False
)

display(outlier_df)

print("\nTotal numerical features:",
      len(numerical_cols))
```

```python

# Convert timestamps into calendar dates
city_data['date'] = city_data['last_updated'].dt.normalize()

# Aggregate multiple observations on the same date
daily_data = (
    city_data.groupby('date')
    .agg(
        temperature=('temperature_celsius', 'mean'),
        precipitation=('precip_mm', 'mean'),
        humidity=('humidity', 'mean'),
        wind_speed=('wind_mph', 'mean')
    )
)

# Create a complete daily date index
full_dates = pd.date_range(
    start=daily_data.index.min(),
    end=daily_data.index.max(),
    freq='D'
)

daily_data = daily_data.reindex(full_dates)
daily_data.index.name = 'date'

# Check missing dates
print("Daily observations:", len(daily_data))
print("Missing temperature dates:",
      daily_data['temperature'].isna().sum())

print("\nMissing dates:")
display(daily_data[daily_data['temperature'].isna()])

print("\nDaily time series:")
display(daily_data.head(10))
```

```python

# Chronological 80:20 train-test split
split_index = int(len(daily_data) * 0.80)

train_data = daily_data.iloc[:split_index].copy()
test_data = daily_data.iloc[split_index:].copy()

print("Training period:")
print(train_data.index.min(), "to", train_data.index.max())

print("\nTesting period:")
print(test_data.index.min(), "to", test_data.index.max())

print("\nTraining observations:", len(train_data))
print("Testing observations:", len(test_data))

print("\nMissing values before handling:")
print("Training:", train_data.isna().sum())
print("Testing:", test_data.isna().sum())

# Fill missing training observations using interpolation
train_data = train_data.interpolate(method='time', limit_area='inside')

print("\nMissing values after training interpolation:")
print(train_data.isna().sum())
```

```python

# 1. Separate training and testing temperature data
train_end = '2024-12-04'

train_temp = daily_data.loc[:train_end, 'temperature'].copy()
test_temp = daily_data.loc['2024-12-05':, 'temperature'].copy()

# Interpolate missing values using training data only
train_temp = train_temp.interpolate(method='time')

# Combine the series for creating historical lag features
temperature_series = pd.concat([train_temp, test_temp])

# 2. Create forecasting features
forecast_data = pd.DataFrame(index=temperature_series.index)

forecast_data['temperature'] = temperature_series

# Previous temperature values
forecast_data['lag_1'] = temperature_series.shift(1)
forecast_data['lag_3'] = temperature_series.shift(3)
forecast_data['lag_7'] = temperature_series.shift(7)

# Previous 7-day average (excluding the current day)
forecast_data['rolling_mean_7'] = (
    temperature_series.shift(1).rolling(window=7).mean()
)

# Calendar features
forecast_data['day_of_year'] = forecast_data.index.dayofyear

forecast_data['month_sin'] = np.sin(
    2 * np.pi * forecast_data.index.month / 12
)

forecast_data['month_cos'] = np.cos(
    2 * np.pi * forecast_data.index.month / 12
)

# Remove rows without sufficient historical information
forecast_data = forecast_data.dropna()

# 3. Split features and target chronologically
features = [
    'lag_1',
    'lag_3',
    'lag_7',
    'rolling_mean_7',
    'day_of_year',
    'month_sin',
    'month_cos'
]

X_train = forecast_data.loc[:train_end, features]
y_train = forecast_data.loc[:train_end, 'temperature']

X_test = forecast_data.loc['2024-12-05':, features]
y_test = forecast_data.loc['2024-12-05':, 'temperature']

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print("Training target observations:", len(y_train))
print("Testing target observations:", len(y_test))
print("Missing training features:", X_train.isna().sum().sum())
print("Missing testing features:", X_test.isna().sum().sum())
```

```python

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

# Initialize models
models = {
    'Linear Regression': LinearRegression(),

    'Random Forest': RandomForestRegressor(
        n_estimators=200,
        max_depth=5,
        random_state=42
    ),

    'Gradient Boosting': GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=42
    )
}

# Train and predict
predictions = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    predictions[name] = model.predict(X_test)

# Persistence baseline: predict today's temperature
# using the previous day's actual temperature
predictions['Persistence Baseline'] = X_test['lag_1'].values

# Evaluation function
def evaluate_model(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mape = np.mean(
        np.abs((actual - predicted) / actual)
    ) * 100

    return mae, rmse, mape

# Compare model performance
results = []

for name, predicted in predictions.items():
    mae, rmse, mape = evaluate_model(
        y_test.values,
        predicted
    )

    results.append({
        'Model': name,
        'MAE': mae,
        'RMSE': rmse,
        'MAPE (%)': mape
    })

results_df = pd.DataFrame(results).sort_values('MAE')

display(results_df.round(3))
```

```python

# Equal-weight ensemble of all four models
ensemble_prediction = np.mean(
    np.column_stack([
        predictions['Persistence Baseline'],
        predictions['Linear Regression'],
        predictions['Random Forest'],
        predictions['Gradient Boosting']
    ]),
    axis=1
)

# Evaluate ensemble
ensemble_mae = mean_absolute_error(y_test, ensemble_prediction)
ensemble_rmse = np.sqrt(
    mean_squared_error(y_test, ensemble_prediction)
)
ensemble_mape = np.mean(
    np.abs((y_test.values - ensemble_prediction) / y_test.values)
) * 100

# Add ensemble to comparison
results_df = pd.concat([
    results_df,
    pd.DataFrame([{
        'Model': 'Equal-Weight Ensemble',
        'MAE': ensemble_mae,
        'RMSE': ensemble_rmse,
        'MAPE (%)': ensemble_mape
    }])
], ignore_index=True)

results_df = results_df.sort_values('MAE').reset_index(drop=True)

display(results_df.round(3))
```

```python

from sklearn.inspection import permutation_importance
import pandas as pd
import matplotlib.pyplot as plt

# 1. Random Forest feature importance
rf_importance = pd.Series(
    models['Random Forest'].feature_importances_,
    index=features,
    name='Random Forest'
)

# 2. Gradient Boosting feature importance
gb_importance = pd.Series(
    models['Gradient Boosting'].feature_importances_,
    index=features,
    name='Gradient Boosting'
)

# 3. Permutation importance for Linear Regression
permutation_result = permutation_importance(
    models['Linear Regression'],
    X_test,
    y_test,
    n_repeats=20,
    random_state=42,
    scoring='neg_mean_absolute_error'
)

permutation_importance_values = pd.Series(
    permutation_result.importances_mean,
    index=features,
    name='Permutation Importance'
)

# 4. Combine all importance results
importance_df = pd.concat(
    [rf_importance, gb_importance, permutation_importance_values],
    axis=1
)

print("Feature Importance Comparison:")
display(importance_df.round(4))

# 5. Plot comparison
importance_df.plot(
    kind='bar',
    figsize=(13, 6)
)

plt.title('Feature Importance Comparison Across Models')
plt.xlabel('Forecasting Features')
plt.ylabel('Importance Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
